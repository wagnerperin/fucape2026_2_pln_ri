```python
import os
import re
import pandas as pd
import numpy as np
from rank_bm25 import BM25Okapi
# 1. Leitura dos decumentos da pasta corpora

dir_corpora = './corpora'
documentos = {}
for i in range(1, 11):
  nome_arq = f"relatorio_{i:02d}.txt"
  with open(os.path.join(dir_corpora, nome_arq), "r", encoding='utf-8') as f:
    documentos[nome_arq] = f.read().strip()

#print(documentos)

# Exibindo uma visão geral do corpus em formato tabular
df_corpus = pd.DataFrame({
    "Arquivo": list(documentos.keys()),
    "Caracteres": [len(doc) for doc in documentos.values()],
    "Conteúdo": list(documentos.values())
})

#display(df_corpus);

# Tokenização e Estatisticas da Coleção

def tokenizar(texto):
  """Normaliza o texto para minúsculos e extra apenas palavras."""
  return re.findall(r'\b\w+\b', texto.lower())

#print(tokenizar("Texto não normalizado e com espaços"))

# Tokenizar todos os documentos do corpus
corpus_tokenizado = [tokenizar(doc) for doc in documentos.values()]
#print(corpus_tokenizado)

# Calculos estatisticos da coleção
doc_lengths = [len(doc) for doc in corpus_tokenizado]
print(doc_lengths)
avg_doc_length = np.mean(doc_lengths)
print(f"Média de Caracteres por documento: {avg_doc_length:.2f}")
vocabulario = sorted(list(set(token for doc in corpus_tokenizado for token in doc)))
print(corpus_tokenizado)
print(vocabulario)

# OKAPI BM25
# 1. Instanciação de dois motores BM25 com calibração distinta de 'b'
bm25_padrao = BM25Okapi(corpus_tokenizado, k1=1.5, b=0.75)
bm25_sem_len = BM25Okapi(corpus_tokenizado, k1=1.5, b=0)

consulta = "rescisão antecipada de contrato e suporte"
q_tokens = tokenizar(consulta)
print(q_tokens)

scores_padrao = bm25_padrao.get_scores(q_tokens)
scores_sem_len = bm25_sem_len.get_scores(q_tokens)

df_comp = pd.DataFrame({
    "Arquivo": list(documentos.keys()),
    "BM25 Padrão": scores_padrao,
    "BM25 Sem Tamanho": scores_sem_len
})

display(df_comp)

# LSA / TruncatedSVD
```
